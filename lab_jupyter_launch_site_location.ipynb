First, let's try to add each site's location on a map using site's latitude and longitude coordinates


The following dataset with the name `spacex_launch_geo.csv` is an augmented dataset with latitude and longitude added for each site.


```python
# Download and read the `spacex_launch_geo.csv`
from js import fetch
import io

URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/spacex_launch_geo.csv'
resp = await fetch(URL)
spacex_csv_file = io.BytesIO((await resp.arrayBuffer()).to_py())
spacex_df=pd.read_csv(spacex_csv_file)
```

Now, you can take a look at what are the coordinates for each site.


```python
# Select relevant sub-columns: `Launch Site`, `Lat(Latitude)`, `Long(Longitude)`, `class`
spacex_df = spacex_df[['Launch Site', 'Lat', 'Long', 'class']]
launch_sites_df = spacex_df.groupby(['Launch Site'], as_index=False).first()
launch_sites_df = launch_sites_df[['Launch Site', 'Lat', 'Long']]
launch_sites_df
```

We first need to create a folium `Map` object, with an initial center location to be NASA Johnson Space Center at Houston, Texas.

We could use `folium.Circle` to add a highlighted circle area with a text label on a specific coordinate. For example,


```python
# Create a blue circle at NASA Johnson Space Center's coordinate with a popup label showing its name
circle = folium.Circle(nasa_coordinate, radius=1000, color='#d35400', fill=True).add_child(folium.Popup('NASA Johnson Space Center'))
# Create a blue circle at NASA Johnson Space Center's coordinate with a icon showing its name
marker = folium.map.Marker(
    nasa_coordinate,
    # Create an icon as a text label
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % 'NASA JSC',
        )
    )
site_map.add_child(circle)
site_map.add_child(marker)
```

```python
# Task 2: Mark the success/failed launches for each site on the map
site_map = folium.Map(location=nasa_coordinate, zoom_start=5)

# Colours
# Success = green, Fail = red
for index, row in spacex_df.iterrows():

    coordinate = [row['Lat'], row['Long']]
    
    if row['class'] == 1:
        color = 'green'
    else:
        color = 'red'

    # CColoured circle
    folium.CircleMarker(
        location=coordinate,
        radius=7,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.7,
        popup=f"Launch Site: {row['Launch Site']}<br>Class: {row['class']}"
    ).add_to(site_map)


site_map
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

*TODO:* Mark down a point on the closest coastline using MousePosition and calculate the distance between the coastline point and the launch site.


```python
# find coordinate of the closet coastline
# e.g.,: Lat: 28.56367  Lon: -80.57163
# distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon
def calculate_distance(lat1, lon1, lat2, lon2):
    
    R = 6373.0  # Earth radius in km

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c

    return distance

# Coordinate launch site
launch_site_lat = 28.562302
launch_site_lon = -80.577356

# Coordinate nearest point on the coast
coastline_lat = 28.56367
coastline_lon = -80.57163

# Distance
distance_coastline = calculate_distance(
    launch_site_lat,
    launch_site_lon,
    coastline_lat,
    coastline_lon
)

print(f"Distance to coastline: {distance_coastline:.2f} km")

# Map
site_map = folium.Map(
    location=[launch_site_lat, launch_site_lon],
    zoom_start=12
)

# Marker launch site
folium.Marker(
    [launch_site_lat, launch_site_lon],
    popup='Launch Site',
    icon=folium.Icon(color='red')
).add_to(site_map)

# Marker coastline point
folium.Marker(
    [coastline_lat, coastline_lon],
    popup=f'Closest Coastline Point<br>Distance: {distance_coastline:.2f} km',
    icon=folium.Icon(color='blue')
).add_to(site_map)

# Line to connect
folium.PolyLine(
    locations=[
        [launch_site_lat, launch_site_lon],
        [coastline_lat, coastline_lon]
    ],
    weight=2,
    color='yellow'
).add_to(site_map)

# Plugin MousePosition
formatter = "function(num) {return L.Util.formatNum(num, 5);};"

mouse_position = MousePosition(
    position='topright',
    separator=' | ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Coordinates:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)

site_map
```

Launch sites are generally in relatively close proximity to coastlines. This is intentional and represents one of the most important characteristics of spaceport selection. Launching rockets near the ocean reduces risks to populated areas because rocket stages and debris can safely fall into the sea during ascent. For example, many SpaceX launch facilities are located along the Florida coast for this reason.

Launch sites also tend to be reasonably close to highways and transport infrastructure. Although launch facilities are isolated for safety reasons, they still require efficient logistics access for:
- transporting rocket components,
- moving fuel and equipment,
- supporting technical teams and operations.

Therefore, highways are usually nearby, but not immediately adjacent to the launch pads.

Railways are sometimes present near launch complexes, but they are generally less critical than highways. Modern aerospace logistics rely more heavily on road transportation and specialized heavy vehicles. In some historical launch facilities, railways were more important for transporting rocket stages and large equipment.

Another important observation is that launch sites are usually located far away from major cities and dense urban areas. This safety buffer minimizes risks associated with:
- launch failures,
- explosions,
- toxic fuel exposure,
- falling debris,
- acoustic shock waves.

For example, Kennedy Space Center and Cape Canaveral are located in relatively isolated coastal areas rather than inside large metropolitan zones.

Overall, the spatial analysis suggests that ideal launch sites balance two competing requirements:
- isolation and safety,
- accessibility and logistical support.
